# Heap Sort

## 📚 Understanding Heap Data Structure

### Why Arrays for Heaps?
In typical tree structures, we use **linked lists** because tree data might not be stored continuously in memory. However, for heaps, we use **arrays** because of a special property.

### 🌳 Complete Binary Tree Requirements

Heap Sort requires the tree to be a **Complete Binary Tree** (or Almost Complete Binary Tree). This means:

#### ✅ Two Critical Prerequisites:

1. **Left-to-Right Insertion**
   - Elements must be inserted from left to right at each level
   - No gaps allowed in any level

2. **Level-by-Level Filling**
   - Before adding any leaf node, the previous level must be completely filled
   - Ensures the tree stays balanced

### 📝 Important Note
> **Complete Binary Tree** = Full Binary Tree (FBT) + Almost Complete Binary Tree (ACBT)
>
> This property allows us to represent the tree structure efficiently using a simple array!

### 🔗 Array Representation Formula
For any node at index `i`:
- **Left Child**: `2*i + 1`
- **Right Child**: `2*i + 2`
- **Parent**: `(i-1) // 2`

## 🌲 Binary Tree Types

### Full Binary Tree (FBT)
A Full Binary Tree is a binary tree in which every node has either **0 or 2 children** (no node has only 1 child).

#### Properties:
- ✅ Every internal node has exactly 2 children
- ✅ All leaf nodes are at the same level (in a perfect FBT)
- ✅ Number of leaf nodes = Number of internal nodes + 1
- ✅ If height is `h`, minimum nodes = `2h + 1`

**Example:**
```
       1
      / \
     2   3
    / \
   4   5
```

---

### Almost Complete Binary Tree (ACBT)
An Almost Complete Binary Tree is a binary tree in which:
- All levels are completely filled except possibly the last level
- The last level is filled from left to right

#### Properties:
- ✅ All levels are full except the last level
- ✅ Last level has nodes as far left as possible
- ✅ Can be efficiently represented as an array
- ✅ If there are `n` nodes, height = `⌊log₂(n)⌋`

**Example:**
```
       1
      / \
     2   3
    / \  /
   4  5 6
```

---

### Complete Binary Tree (CBT)
A Complete Binary Tree combines properties of both FBT and ACBT.

#### Properties:
- ✅ All levels are completely filled except possibly the last
- ✅ Last level is filled from left to right (ACBT property)
- ✅ Can have nodes with 0, 1, or 2 children
- ✅ **Perfect for Heap implementation** 🎯
- ✅ Height is always `O(log n)`

**Example:**
```
       1
      / \
     2   3
    / \ / \
   4  5 6  7
  /
 8
```
